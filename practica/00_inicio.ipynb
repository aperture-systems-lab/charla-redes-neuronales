## **Parte 1:** El entorno

Usando `uv sync` en la terminal se instalan todas las dependencias. Después abre la carpeta en VS Code con `code .` y elige el kernel de `practica/.venv` en el botón de arriba llamado `Select Kernel`.

El proyecto se maneja con [**uv**](https://docs.astral.sh/uv/). Tres archivos lo definen.

- **`pyproject.toml`** declara las dependencias. librerías, versión de python y otras cosas del estilo.
- **`uv.lock`** fija la versión exacta de cada paquete. 
- **`.venv/`** es la instalación. Se puede borrar y se reconstruye con `uv sync`.
